# Algèbre de Boole

## Origines historiques

En 🇬🇧 **George Boole**, mathématicien britannique, invente au XIXe siècle un formalisme destiné à exprimer des raisonnements logiques : **l'Algèbre de Boole**. À cette époque, la notion d'informatique n’existe pas encore, bien que des machines effectuant des calculs automatisés soient déjà en usage — comme la **Pascaline** de Blaise Pascal.

Bien plus tard, au XXe siècle, 🇺🇸 **Claude Shannon**, ingénieur et mathématicien américain, démontre que des circuits électriques peuvent être utilisés pour résoudre tous les problèmes formulables à l'aide de l'algèbre de Boole. Ses travaux marquent une avancée majeure dans la naissance de l'informatique.

Pendant la Seconde Guerre mondiale, les contributions de 🇬🇧 **Alan Turing** et de 🇺🇸🇭🇺 **John von Neumann** poseront les bases théoriques et pratiques de l’**informatique moderne**.

---

## Qu'est-ce que l'Algèbre de Boole ?

L'**Algèbre de Boole**, également appelée **calcul booléen**, est une branche des mathématiques qui permet de représenter la logique sous une forme algébrique.

Elle repose sur un **ensemble contenant uniquement deux éléments**, appelés **valeurs booléennes** :

- `F` ou `False` (FAUX)
- `V` ou `True` (VRAI)

---

## Opérations booléennes

L’algèbre de Boole permet de manipuler ces valeurs à l’aide :

- **d’opérateurs booléens**, qui agissent sur un ou deux booléens
- ou **d’expressions/fonctions booléennes**, qui généralisent les opérations à plusieurs booléens

Un **opérateur** est un symbole représentant une opération entre un ou plusieurs éléments.

### Les opérations fondamentales sont :

- **Conjonction** : `ET` (AND)
- **Disjonction** : `OU` (OR)
- **Négation** : `NON` (NOT)

## Tables de vérité

Une **table de vérité** décrit le comportement d'une opération booléenne en fonction de toutes les combinaisons possibles de ses entrées.

### Négation (NON / NOT)

| x   | NON x |
|-----|-------|
| F   | V     |
| V   | F     |

---

### Conjonction (ET / AND)

| x   | y   | x ET y |
|-----|-----|--------|
| F   | F   | F      |
| F   | V   | F      |
| V   | F   | F      |
| V   | V   | V      |

---

### Disjonction (OU / OR)

| x   | y   | x OU y |
|-----|-----|--------|
| F   | F   | F      |
| F   | V   | V      |
| V   | F   | V      |
| V   | V   | V      |

---

## Exemples d'expressions booléennes

Voici quelques expressions courantes, leur signification et leur évaluation.

### Exemple 1 :  
**Expression** : `x ET (NON y)`  
**Interprétation** : x est vrai **et** y est faux

| x   | y   | NON y | x ET (NON y) |
|-----|-----|--------|--------------|
| F   | F   | V      | F            |
| F   | V   | F      | F            |
| V   | F   | V      | V            |
| V   | V   | F      | F            |

---

### Exemple 2 :  
**Expression** : `(x OU y) ET (NON x)`  
**Interprétation** : x ou y est vrai, **et** x est faux

| x   | y   | x OU y | NON x | Résultat |
|-----|-----|--------|--------|-----------|
| F   | F   | F      | V      | F         |
| F   | V   | V      | V      | V         |
| V   | F   | V      | F      | F         |
| V   | V   | V      | F      | F         |

## Symboles des opérations booléennes

| Opération | Nom       | Symboles courants                      | Utilisation / Remarques                                |
|-----------|-----------|----------------------------------------|---------------------------------------------------------|

|           |           |                                        | - `and` en Python                                       |

|           |           |                                        | - `or` en Python                                        |
| NON       | NOT       | ¬, !, `not`                            | - ¬ en logique mathématique                             |
|           |           |                                        | - ! en programmation (C, Java, etc.)                    |
|           |           |                                        | - `not` en Python                                       |

```python
# Déclaration de deux variables booléennes
x = True
y = False

# Opérateur AND (ET)
print("x and y =", x and y)   # Résultat : False, car True ET False = False
```

```python
# Déclaration de deux variables booléennes
x = True
y = False

# Opérateur OR (OU)
print("x or y =", x or y)     # Résultat : True, car True OU False = True
```

```python
# Déclaration de deux variables booléennes
x = True
y = False

# Opérateur NOT (NON)
print("not x =", not x)       # Résultat : False, car NON True = False
print("not y =", not y)       # Résultat : True, car NON False = True
```

```python
# Déclaration de deux variables booléennes
x = True
y = False
# Combinaisons plus complexes
print("x and not y =", x and not y)  # Résultat : True, car True ET NON False = True

print("(x or y) and not x =", (x or y) and not x)  
# Résultat : False, car (True OU False) ET NON True = True ET False = False
```
