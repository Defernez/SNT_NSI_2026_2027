## 4. Théorème de Pythagore : <br> Définissez la fonction hypotenuse.<br>
Elle acceptera deux paramètres notés a et b correspondant aux longueurs des 2 plus petits côtés d'un triangle rectangle (exprimées avec la même unité).
La valeur de retour de cette fonction sera la longueur de l'hypoténuse.
<code>
\>\>\> hypotenuse(2, 3)
3.605551275463989
\>\>\> hypotenuse(5, 2)
5.385164807134504

</code>

<h2>🧠 Rappel mathématique : Théorème de Pythagore</h2>
<p>
  Dans un <strong>triangle rectangle</strong>, l’hypoténuse (le plus grand côté, opposé à l’angle droit)
  se calcule à partir des deux autres côtés (appelés les <strong>côtés de l’angle droit</strong>) grâce à cette formule :
</p>

<p style="font-size: 1.2em; font-family: monospace;">
  <strong>hypoténuse</strong> = √(a² + b²)
</p>


```python
#Le code ci-dessous importe le module mathématique de python
#math.sqrt() est la fonction qui retourne une racine carrée
#rappel ** permet d'exprimer une puissance
import math

math.sqrt(2**2 + 3**2)
```
