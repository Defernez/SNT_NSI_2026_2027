<h1>Itérations sur les tableaux

En Python, il existe plusieurs façons d’itérer sur les tableaux (plus précisément, les listes). Voici un aperçu des méthodes les plus courantes pour itérer sur une liste :

```python
# Boucle for
chifoumi = ["pierre", "feuille", "ciseaux"]

for action in chifoumi:
    print(action)
```

```python
# Avec range() et les indices
chifoumi = ["pierre", "feuille", "ciseaux"]

for i in range(len(chifoumi)):
    print(chifoumi[i])
```

```python
# Avec enumerate() (utile pour l'indice et la valeur)
chifoumi = ["pierre", "feuille", "ciseaux"]

for i, action in enumerate(chifoumi):
    print(i, end ="")
    print(" -> " , action)
```

```python
# Avec while
cours = ["Math", "NSI", "SVT"]
i = 0

while i < len(cours):
    print(cours[i])
    i += 1
```

```python
# Compréhensions de liste 
fruits = ["pomme", "banane", "fraise"]
[print(fruit) for fruit in fruits]  # Ce n'est pas très courant pour juste imprimer, retourne une liste de None
```
