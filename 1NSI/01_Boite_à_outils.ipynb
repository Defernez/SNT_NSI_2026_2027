## 🧩 Types de cellules

Dans Jupyter, il existe plusieurs types de cellules :

- **Markdown** :  
  Ce type de cellule permet de rédiger du texte, un peu comme sur un site web.  
  Tu peux utiliser le langage *Markdown* (un langage de balisage léger) et même y intégrer du **HTML**.  
  Les deux sont compatibles dans Jupyter.

- **Texte brut** :  
  Tu écris du texte sans aucune mise en forme. C’est juste du texte simple.

- **Code** :  
  Ce type de cellule est utilisé pour écrire du code Python.  
  Lorsque tu appuies sur le bouton **Exécuter (|>)**, le code s’exécute et reste en mémoire tant que :
  - tu ne redémarres pas le noyau,
  - ou tu ne recharges pas la page.

| Opération            | Opérateur | Type du résultat       |
