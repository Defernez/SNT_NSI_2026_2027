<h1>La boucle non bornée WHILE

Pour fonctionner, un ordinateur a besoin d’un processeur. Ce processeur utilise de la mémoire, dans laquelle est stocké un code. Ce code est composé d’opérations, que nous appellerons des opcodes, et lorsqu’il est lu et exécuté par le processeur, il déclenche les actions qui constituent le programme.

Le processeur exécute simplement les opcodes un par un. Une fois tous les opcodes exécutés, le processeur cesse son activité, et le programme se termine.

L’un des principaux programmes chargés en mémoire au démarrage est le système d’exploitation. Celui-ci contient généralement une boucle principale qui tourne en continu, empêchant le système de s’arrêter tant que l’utilisateur ne le décide pas. Cette boucle est souvent appelée boucle while (tant que), et elle reste active pour surveiller les événements du système.

Durant cette boucle, le système va détecter de nombreuses choses, comme les mouvements de la souris, les clics, ou encore les touches du clavier. Ce sont ce qu’on appelle des interruptions, qui déclenchent des réactions dans le programme.

Il faut programmer cette boucle avec soin, car si elle est mal conçue, elle peut tourner en rond indéfiniment sans accomplir quoi que ce soit d’utile. Dans certains langages comme Python, un chien de garde (watchdog) ou des mécanismes de sécurité peuvent interrompre un programme s’il tourne trop longtemps sans raison apparente, évitant ainsi les blocages ou les fuites de ressources.

![img](https://assets.rockpapershotgun.com/images/2019/07/AMD-Ryzen-5-2600-best-budget-gaming-cpu.jpg)

```python
compteur = 10
while compteur > 3:
    print(compteur)
    compteur = compteur - 1
```

⚠️ Attention aux boucles infinies

Les boucles while en Python peuvent parfois ne jamais s'arrêter. On appelle cela une boucle infinie.
Cela arrive quand la condition après while reste toujours vraie : la boucle recommence encore et encore, sans fin. Dans ce cas, il faut forcer l’arrêt du programme (en fermant l’exécution).

🧠 Conseil :

Vérifie toujours que quelque chose change dans la boucle pour que la condition puisse devenir fausse un jour.

```python
#Exemple 1 : boucle avec condition toujours vraie,
# vous devrez relancer le noyau sous peine d'obtenir une boucle infinie
# ou mieux, ne pas exécuter ce code :-)
while True:
    print("Bonjour !")
   
```

```python
#Exemple 2: variable qui ne rapproche jamais de la fin
#Idem, ne s'arrête jamais
i = 5
while i > 3:
    i = i + 1
```

```python
# Capital de départ
capital_depart = 100

# Gain que l'on souhaite obtenir
gain = 20

# Calcul du capital à atteindre
capital_cible = capital_depart + gain

# Initialisation des variables
capital = capital_depart
annee = 0

# Boucle : tant que le capital n'a pas atteint le capital cible
while capital < capital_cible:
    capital = capital * 1.05  # augmentation de 5%
    annee = annee + 1         # on compte les années

# Affichage du résultat
print("Il faut", annee, "années pour obtenir au moins", gain, "€ de gain.")
```

🛑 L’instruction break : sortir d’une boucle avant la fin

Dans une boucle while, on peut utiliser break pour interrompre la boucle immédiatement, même si la condition de la boucle est encore vraie.

```python
i = 1
while i < 5:
    print(i)
    if i == 2:
        break
    i = i + 1

print("Fin")
```

```python
while True:
    a = input("saisir quelque chose ou 'q' pour quitter")

    if a == "q":
        print("Fin")
        break 
    else:
        print("echo", a)
```

🔁 L'instruction continue en Python

L’instruction continue permet de sauter le reste du code d’une boucle et de passer directement au tour suivant.
Contrairement à break, elle ne quitte pas la boucle, elle revient simplement au début de l’itération suivante.

👉 On peut l’utiliser dans une boucle while ou dans une boucle for.

```python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```
