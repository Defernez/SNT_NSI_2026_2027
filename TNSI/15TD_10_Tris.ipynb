<h2>Le tri par sélection

https://youtu.be/8u3Yq-5DTN8

Sur une liste de <em>n</em> éléments (numérotés de 0 à <em>n-1 , attention un tableau de 5 valeurs (5 cases) sera numéroté de 0 à 4</em> et non de 1 à 5), le principe du tri par sélection est le suivant :</p>
<p dir="auto" data-sourcepos="7:1-7:34">Pour <em>i</em> allant de 0 à  <em>n</em> - 2 :</p>
<ul dir="auto" data-sourcepos="9:1-12:0">
<li data-sourcepos="9:1-10:0">
<p data-sourcepos="9:3-9:140">rechercher le plus petit élément de la partie de la liste allant de l'indice <em>i</em> à l'indice <em>n</em> - 1 en utilisant un comparateur adapté</p>
</li>
<li data-sourcepos="11:1-12:0">
<p data-sourcepos="11:3-11:44">l'échanger avec l'élément d'indice <em>i</em>.</p>
</li>
</ul>
<p dir="auto" data-sourcepos="13:1-13:85"><strong>Exemple : Considérons la liste  8, 5, 2, 6, 9, 3, 1, 4, 0, 7 de taille <em>n</em> = 10.</strong></p>
<ul dir="auto" data-sourcepos="15:1-17:0">
<li data-sourcepos="15:1-15:54">Ses éléments son numérotées de 0 à <em>n</em> - 1 = 9.</li>
<li data-sourcepos="16:1-17:0">Le comparateur est <code>&lt;</code>
</li>
</ul>
<p dir="auto" data-sourcepos="18:1-18:17"><em>Etape 1 : i = 0</em></p>
<ul dir="auto" data-sourcepos="20:1-26:0">
<li data-sourcepos="20:1-21:0">
<p data-sourcepos="20:3-20:102">le plus petit élément de la liste entre les indice <em>i</em> = 0 et <em>n</em> - 1 = 9 est 0, placé au rang 7.</p>
</li>
<li data-sourcepos="22:1-26:0">
<p data-sourcepos="22:3-22:47">on échange donc le rang i = 0 avec le rang 7</p>
<p data-sourcepos="24:2-24:48"><strong>On obtient  0, 8, 5, 2, 6, 9, 3, 1, 4, 8, 7</strong></p>
</li>
</ul>

<ul dir="auto" data-sourcepos="29:1-36:0">
<li data-sourcepos="29:1-30:0">

<li data-sourcepos="31:1-36:0">
<p data-sourcepos="31:3-31:44">on échange donc le rang 1 avec le rang 7.</p>

<h2 dir="auto" data-sourcepos="101:1-101:28">
<a aria-hidden="true" href="#terminaison-et-correction" class="anchor" id="user-content-terminaison-et-correction"></a>Terminaison et correction</h2>
<ul dir="auto" data-sourcepos="103:1-108:0">
<li data-sourcepos="103:1-105:0">
<p data-sourcepos="103:3-103:569">La <strong>terminaison</strong> de l'algorithme est le fait que l'algorithme se termine bien et, donc, ne boucle sans fin. Ce qui nous permet d'être  certain de la <strong>terminaison</strong>  est appelé <strong>variant de boucle</strong>. Il s'agit d'un nombre entier qui décroit strictement et garantit la sortie de la boucle lorsqu'il atteint 0. Ici le <strong>variant de boucle</strong> est la longueur de la liste qui reste à trier : cette longueur est un nombre entier qui diminue de 1 à chaque tour et atteindra donc fatalement 0 en sonnant la fin du tri en garantissant la <strong>terminaison</strong> de l'algorithme.</p>
</li>
<li data-sourcepos="106:1-108:0">
<p data-sourcepos="106:3-106:408">La <strong>correction</strong> de cette algorithme est l'assurance qu'il réalise bien ce pourquoi il est fait. Ce qui garantit cette correction est appelée <strong>invariant de boucle</strong> : c'est une propriété qui doit rester vraie du début à la fin de l'algorithme. Ici, notre <strong>invariant de boucle</strong> est le fait que la partie de la liste comprise entre l'élément 0 et l'élément <em>i</em> est toujours correctement triée.</p>
</li>
</ul>
<h2 dir="auto" data-sourcepos="109:1-109:24">
<a aria-hidden="true" href="#co%C3%BBt-de-lalgorithme" class="anchor" id="user-content-coût-de-lalgorithme"></a>Coût de l'algorithme</h2>
<p dir="auto" data-sourcepos="111:1-111:232">Trier une liste de n éléments nécessite donc n - 1 étape. Mais, chaque étape nécessite un certain nombre de comparaison. C'est ce nombre qui détermine l'efficacité du tri : plus il est faible, plus l'algorithme est efficace.</p>
<p dir="auto" data-sourcepos="115:1-115:66"><strong>Reprenons notre exemple :  la liste  6, <em>5</em>, 3, 1, 8, 7, 2, 4</strong></p>
<ul dir="auto" data-sourcepos="117:1-146:0">
<li data-sourcepos="117:1-120:0">
<p data-sourcepos="117:4-117:95">On commence par comparer 5 avec 6 ce qui conduit à placer 5 à gauche du 6 -&gt; 1 comparaison</p>
<p data-sourcepos="119:2-119:44">​	<strong>On obtient 5, 6, <em>3</em>, 1, 8, 7, 2, 4</strong></p>
</li>
<li data-sourcepos="121:1-124:0">
<p data-sourcepos="121:3-121:81">On compare 3 à 6 puis 3 à 5 pour placer 3 en tête de liste -&gt; 2 comparaisons</p>
<p data-sourcepos="123:2-123:44">​	<strong>On obtient 3, 5, 6, <em>1</em>, 8, 7, 2, 4</strong></p>
</li>
<li data-sourcepos="125:1-128:0">
<p data-sourcepos="125:3-125:75">On continue en comparant 1 à 6 puis 1 à 5 puis 1 à 3 -&gt; 3 comparaisons</p>
<p data-sourcepos="127:2-127:45">​	<strong>On obtient 1, 3, 5, 6,  <em>8</em>, 7, 2, 4</strong></p>
</li>
<li data-sourcepos="129:1-132:0">
<p data-sourcepos="129:3-129:110">On compare 8 à 6 : <strong>8 est à sa place, inutile de continuer !!</strong> -&gt; une seule comparaison et rien ne bouge</p>
<p data-sourcepos="131:2-131:41"><strong>On obtient 1, 3, 5, 6,  8, <em>7</em>, 2, 4</strong></p>
</li>
<li data-sourcepos="133:1-136:0">
<p data-sourcepos="133:3-133:101">On compare 7 à 8 puis 7 à 6 : <strong>7 est à sa place, inutile de continuer !!</strong> -&gt; deux comparaisons</p>
<p data-sourcepos="135:2-135:42"><strong>On obtient 1, 3, 5, 6,  7, 8 , <em>2</em>, 4</strong></p>
</li>
<li data-sourcepos="137:1-140:0">
<p data-sourcepos="137:3-137:99">On compare 2 à 8 puis 2 à 7, puis  2 à 6 puis 2 à 5 puis 2 à 3 puis 2 à 1 -&gt; 6 comparaisons</p>
<p data-sourcepos="139:2-139:40"><strong>On obtient 1, 2, 3, 5, 6, 7, 8, <em>4</em></strong></p>
</li>
<li data-sourcepos="141:1-146:0">
<p data-sourcepos="141:3-141:111">On compare 4 à 8 puis 4 à 7 puis 4 à 6 puis 4 à 5 puis 4 à 3 : <strong>4 est à sa place !</strong> -&gt; 5 comparaisons</p>
<p data-sourcepos="143:2-143:50"><strong>On obtient 1,2,3,4,5,6,7,8 : c'est terminé !</strong></p>
</li>
</ul>
<p dir="auto" data-sourcepos="147:1-147:65">Nous avons effectué 1 + 2 + 3 + 1 + 2 + 6 + 5 = 20 comparaisons</p>
<p dir="auto" data-sourcepos="151:1-151:242">Lors d'un <a href="/JB_info/nsi_tale/-/blob/master/algorithmique/rappels_premiere/tri_selection.md">tri par sélection</a>, nous avons vu que le nombre de comparaisons pour une liste de taille 8 serait $$ \frac {8 * 7}{2} = 28$$  , c'est à dire <strong>8 de plus que le tri par insertion qui est donc plus performant !!</strong></p>
<p dir="auto" data-sourcepos="154:1-154:292">Cela s'explique par le fait que le nombre de comparaisons nécessaires avec le tri par sélection est toujours le même alors que, avec le tri par insertion, si un élément est à sa place, c'est à dire si la liste est déjà un peu triée, le  nombre de comparaisons nécessaires diminue !</p>
<blockquote dir="auto" data-sourcepos="157:1-164:1">
<p data-sourcepos="157:3-157:236">Lors du tri par insertion d'une liste de n éléments, le nombre de comparaison est au maximum de <span data-math-style="inline"><span class="katex"><span class="katex-mathml"><math xmlns="http://www.w3.org/1998/Math/MathML"><semantics><mrow><mfrac><mrow><mi>n</mi><mo stretchy="false">(</mo><mi>n</mi><mo>−</mo><mn>1</mn><mo stretchy="false">)</mo></mrow><mn>2</mn></mfrac></mrow><annotation encoding="application/x-tex"> $$ \frac {n ( n-1)}{2} $$ </annotation></semantics></math></span><span class="katex-html" aria-hidden="true"><span class="base"><span class="strut" style="height:1.355em;vertical-align:-0.345em;"></span><span class="mord"><span class="mopen nulldelimiter"></span><span class="mfrac"><span class="vlist-t vlist-t2"><span class="vlist-r"><span class="vlist" style="height:1.01em;"><span style="top:-2.6550000000000002em;"><span class="pstrut" style="height:3em;"></span><span class="sizing reset-size6 size3 mtight"><span class="mord mtight"><span class="mord mtight">2</span></span></span></span><span style="top:-3.23em;"><span class="pstrut" style="height:3em;"></span><span class="frac-line" style="border-bottom-width:0.04em;"></span></span><span style="top:-3.485em;"><span class="pstrut" style="height:3em;"></span><span class="sizing reset-size6 size3 mtight"><span class="mord mtight"><span class="mord mathnormal mtight">n</span><span class="mopen mtight">(</span><span class="mord mathnormal mtight">n</span><span class="mbin mtight">−</span><span class="mord mtight">1</span><span class="mclose mtight">)</span></span></span></span></span><span class="vlist-s">​</span></span><span class="vlist-r"><span class="vlist" style="height:0.345em;"><span></span></span></span></span></span><span class="mclose nulldelimiter"></span></span></span></span></span></span> (quand la liste est à 100% désordonnée) et au minimum de <em>n - 1</em> si la liste est déjà totalement ordonnée.</p>
<p data-sourcepos="159:3-159:224">Le tri par insertion est donc <strong>au pire</strong> équivalent en terme de performance au tri par sélection mais, la plupart du temps, dès que quelques éléments sont au départ déjà près de leur place, il sera bien meilleur.</p>
<p data-sourcepos="163:3-163:216">L'ordre de grandeur du nombre de comparaison est au pire  <em>n²</em>  et au mieux <em>n</em>: on dit que sa <strong>complexité est au pire quadratique</strong> (aussi noté  <strong><em>O(n²)</em></strong>) et au mieux <strong>linéaire</strong> (aussi noté <strong><em>O(n)</em></strong>)</p>
