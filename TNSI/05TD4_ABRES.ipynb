<h2 dir="auto" data-sourcepos="5:1-5:17">
<a aria-hidden="true" href="#i-d%C3%A9finition" class="anchor" id="user-content-i-définition"></a>I. Définition</h2>

<p dir="auto" data-sourcepos="7:1-7:240">En informatique nous parlerons d'un type de graphe particulier : les <em>arbres enracinés</em> ou <em>arborescence</em>. Par abus de langage on omettra souvent le terme « enraciné » et nous parlerons simplement d'<strong>arbre</strong>. En voici une définition :</p>

<p dir="auto" data-sourcepos="9:1-9:85">Un arbre (enraciné) ou une arborescence est un <strong>graphe orienté sans cycle</strong> dont :</p>

<ul dir="auto" data-sourcepos="11:1-13:0">
<li data-sourcepos="11:1-11:65">un seul sommet n'a aucun parent, on l'appellera la <strong>racine</strong> ;</li>
<li data-sourcepos="12:1-13:0">tous les autres sommets n'ont qu'un seul parent.</li>
</ul>

<p dir="auto" data-sourcepos="14:1-14:57">Cette définition fut établie par Arthur Cayley en 1889.</p>

<ul dir="auto" data-sourcepos="61:1-64:0">

<li data-sourcepos="62:1-62:84">la <strong>hauteur</strong> d'un arbre est la plus grande profondeur d'une feuille de l'arbre ;</li>
<li data-sourcepos="63:1-64:0">la <strong>taille</strong> d'un arbre est le nombre de nœuds.</li>
</ul>

<h3 dir="auto" data-sourcepos="125:1-125:32">
<a aria-hidden="true" href="#3-encadrement-de-la-hauteur" class="anchor" id="user-content-3-encadrement-de-la-hauteur"></a>3. Encadrement de la hauteur</h3>

Il est possible d'avoir des arbres binaires de même taille, mais de "forme" très différente :

![abr1](https://framagit.org/JB_info/nsi_tale/-/raw/master/structures_de_donnees/img/filiforme.png) ![abr2](https://framagit.org/JB_info/nsi_tale/-/raw/master/structures_de_donnees/img/complet.png)

<p dir="auto" data-sourcepos="131:1-131:332">Sur le schéma ci-dessus le premier arbre est dit <strong>filiforme</strong> alors que le deuxième est dit <strong>complet</strong> (un arbre binaire est complet si  tous les nœuds possèdent 2 fils et toutes les feuilles se situent à la même profondeur). On pourra aussi dire que le premier arbre est déséquilibré alors que le second est équilibré.</p>

<p dir="auto" data-sourcepos="133:1-133:200">Si on prend un arbre filiforme de taille <code>n</code>, on peut dire que la hauteur de cet arbre est égale à <code>n−1</code>. Dans le cas de l'arbre filiforme ci-dessus, nous avons <code>n=7</code> et donc la hauteur est 7-1=6.</p>

Si on prend un arbre complet de taille `n`, on peut démontrer que la hauteur de cet arbre est égale à la partie entière de $log_2(n)$.

Dans le cas de l'arbre complet ci-dessus, nous avons :

$$
log_2(7) = 2,8
$$

Donc, en prenant la partie entière, on a bien la hauteur de l'arbre égale à `2`.


<p dir="auto" data-sourcepos="137:1-137:121">Un arbre filiforme et un arbre complet étant deux cas extrêmes, on peut affirmer que pour un arbre binaire quelconque :</p>

$$
\lfloor \log_2(n) \rfloor \leq hauteur \leq n - 1
$$

(avec `n` la taille de l'arbre, et $\lfloor \log_2(n) \rfloor$ permet de prendre la partie entière du logarithme base 2 de `n`).
