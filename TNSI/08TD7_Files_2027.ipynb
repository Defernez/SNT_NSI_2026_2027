<p dir="auto" data-sourcepos="9:1-9:193">La <strong>file</strong> est une structure de données qui fonctionne sur le principe du <strong>premier arrivé premier sorti</strong>. Pour une file, on parle de mode <strong>FIFO</strong> pour <strong>F</strong>irst <strong>I</strong>n, <strong>F</strong>irst <strong>O</strong>ut.</p>
<p dir="auto" data-sourcepos="13:1-13:78"><a rel="nofollow noreferrer noopener" href="https://upload.wikimedia.org/wikipedia/commons/2/2a/FIFO_PEPS.png" class="no-attachment-icon" target="_blank"><img class="js-lazy-loaded qa-js-lazy-loaded" decoding="async" alt="une file" src="https://upload.wikimedia.org/wikipedia/commons/2/2a/FIFO_PEPS.png" loading="lazy"></a></p>
<p dir="auto" data-sourcepos="17:1-17:52">Par analogie, on peut penser à une file d'attente :</p>
<ul dir="auto" data-sourcepos="19:1-21:0">
<li data-sourcepos="19:1-19:49">les gens entrent dans la file par l'arrière.</li>
<li data-sourcepos="20:1-21:0">le premier qui entre dans la file sera le premier à en sortir</li>
</ul>
<p dir="auto" data-sourcepos="22:1-22:65"><strong>On est bien sur le principe du premier arrivé, premier sorti</strong></p>

<h2 dir="auto" data-sourcepos="5:1-5:34">
<a aria-hidden="true" href="#i-interface-vs-impl%C3%A9mentation" class="anchor" id="user-content-i-interface-vs-implémentation"></a>I. Interface VS Implémentation</h2>
<p dir="auto" data-sourcepos="7:1-7:153">Pour chaque structure de données linéaire (tableau, liste, pile, file), nous avons vu d'abord son interface puis au moins un exemple d'implémentation.</p>
<p dir="auto" data-sourcepos="9:1-9:268">L'<strong>interface</strong> d'une structure de données définit la façon dont on interagit avec cette structure. Cette interface est composée d'un ensemble de méthodes primitives. On manipule alors un type abstrait de données, qui n'existe pas physiquement dans l'ordinateur.</p>
<p dir="auto" data-sourcepos="11:1-11:417">Afin de rendre la structure utilisable par une machine, il faut l'implémenter dans un langage de programmation. Cette <strong>implémentation</strong> dépend du langage utilisé mais également de ce qu'on souhaite faire avec la structure : selon l'implémentation choisie, les méthodes de la liste seront plus ou moins efficaces et bien choisir l'implémentation rendra l'algorithme qui l'utilisera plus ou moins <em>performant</em>.</p>
