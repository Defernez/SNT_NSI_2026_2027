# II. Rappel sur les docstrings en Python

Nous faisons une petite pause dans l'évolution de notre programme pour parler des docstrings. Les docstrings sont la documentation de votre programme. Ils ne changent absolument pas le fonctionnement du programme mais permettent aux utilisateurs et développeurs de votre programme de mieux le comprendre.
La documentation est souvent négligée par les développeurs car on veut souvent aller vite et nous avons l'impression que cela ne sert à rien. Erreur ! La documentation permet d'éviter de nombreux bugs car le code est plus compréhensible. De plus, il est fort probable que ça soit vous qui essayerez de comprendre ce code dans plusieurs semaines ou mois. La documentation vous permettra de gagner beaucoup de temps.<br>
En effet, lorsqu'on écrit un programme tout est frais et logique dans notre tête, mais plusieurs mois après, il peut être difficile de comprendre notre propre code sans explications. Pour une personne extérieure cela peut même être impossible. Voilà pourquoi il est indispensable de documenter son code avec des commentaires et des docstrings !
Que cela soit un rappel ou non, voici le fonctionnement des docstrings.<br>
<br>
Qu'est-ce qu'une docstring ?<br>
Une docstring est une chaîne de caractères avec trois doubles quotes """ situées au début d'un module, d'une fonction, d'une classe (on verra cela plus tard) ou d'une méthode (on verra aussi cela plus tard). Elle explique ce que fait le code et donne ce que retourne une fonction ou une méthode.
Une docstring s'écrit ainsi :<br>
<br>
    . elle est encadrée par trois apostrophes ''' (ou bien par trois guillemets """)<br>
    . la première ligne donne une brève description de ce que fait la fonction<br>
    . pour chaque paramètre, on crée une ligne qui contient :param nom_du_parametre: (type du paramètre) suivi d'une brève description du paramètre<br>
    . pour le résultat on crée une ligne qui contient :return: (type du résultat) suivi d'une brève description du résultat
<br><br>
Regardons donc un exemple pour la fonction combat :

```python
def combat(p1, p2):
    """
    Effectue un combat entre deux personnages et renvoie le nom du vainqueur.
    
    :param p1: (dict) le premier personnage
    :param p2: (dict) le second personnage
    :return: (str) le nom du vainqueur
    """
    pass
```

Cette chaîne de caractères sera ignorée lors de l'exécution du programme. Par contre elle sera utilisée lorsqu'on demande de l'aide sur une fonction avec help().

# III. Rappel sur les tests en Python

## Pourquoi faire des tests ?

Un test permet de dire à Python le résultat attendu d'une fonction (ou autre) pour des paramètres donnés. Cela permet de vérifier qu'une fonction fait bien ce qui lui est demandé. C'est particulièrement utile, lorsqu'on modifie légèrement un programme complexe, pour s'assurer que le reste du code n'est pas « cassé » suite à ce changement. Ainsi, on lance les tests et si tout fonctionne, on peut raisonnablement penser que nous n'avons pas ajouté d'erreurs.

## Comment faire des tests ?

Il existe plusieurs façons de faire des tests en Python. On peut utiliser des bibliothèques telles de pytest ou unitest. Ces deux bibliothèques sont assez puissantes mais demandent des connaissances un peu plus avancées que les doctests que nous allons utiliser.
Les doctests s'intègrent dans les docstrings de Python. Même si cela peut paraître déroutant au début, cela reste la méthode la plus simple pour effectuer des tests en Python. Pour comprendre commençons par un exemple :

```python
def somme(a, b):
    """
    # Ici c’est la documentation de la fonction
    Ajoute deux nombres.
    
    :param a: (int) un nombre
    :param b: (int) un second nombre
    :return: (int) la somme de a et b
    
    # Le test avec son résultat
    >>> somme(2, 2)
    4
    """
    return a + b

# On demande à Python de réaliser les tests
if __name__ == "__main__":
    import doctest
    doctest.testmod()
```

La programmation procédurale est un paradigme, c'est-à-dire une famille de programmation, qui repose sur l’utilisation de procédures qui rendent le code réutilisable. Ce paradigme permet de rendre le programme plus clair et plus simple. On peut alors décomposer en procédures et en fonctions un programme qui semble complexe au premier abord.
Il existe un paradigme dont le but est de coder le programme d’une façon plus intuitive à l’aide d’objets. Ainsi les objets sont plus facile à comprendre et plus proche du raisonnement humain. La programmation orienté objet (POO) utilise donc des objets interconnectés entreeux comme des briques de lego (briques logiciels).


La POO consiste à utiliser et donc à créer des objets.
Les objets possèdent des attributs et des méthodes qui sont liés à leurs classes.
Les attributs font la liaison avec les variables contenus dans l’objet.
Les attributs peuvent contenir des données.
Les méthodes sont des fonctions.
Les méthodes sont à l’intérieur de la classe.
La classe définit l’objet lui-même.
Chaque objet est une instance d’une classe, c'est-à-dire qu’avec l’aide d'une classe on peut créer une multitude d’objets. L’objet hérite des méthodes et des attributs de sa classe.
La classe peut être vue comme un moule faisant partie d’une usine qui fabrique des objets.
Les caractéristiques de l’objet sont renseignées à la classe (l’usine) par l’intermédiaire des attributs, ensuite, l’objet applique les méthodes qu’il possède.
Lorsqu’une classe fabrique un objet on dit qu’elle l’instancie.
Chaque objet est unique et possède sa propre adresse en mémoire, il peut avoir des attributs différents.


Self représente donc l'objet qui sera stoqué dans la mémoire à une adresse unique et personnelle. Self est donc notre objet et contient toutes les variables ou attributs et on peux y accèder. On peut imaginer que self est un sac à dos et il contient tout ce que notre objet a besoin. La classe elle même, qui peut être représenté par une usine qui fabrique les sacs à dos 'self' et les remplits de méthodes et d'attributs, possède sa propre adresse dans la mémoire et bien différente de celle des ses objets fils ou ses sacs à dos fils. Ainsi accèder à une variable (ou attribut) de classe, c'est accèder à la classe elle même, dans sa zone mémoire. Modifier une variable de classe n'affecte pas les objets déjà créés.

```python
# exemple de classe en une seule cellule

class Eleve:
    def __init__(self, nom, prenom, age):
        self.nom=nom
        self.prenom=prenom
        self.age=age
        
    def presentation(self):
        print("bonjour, je m'apelle " + self.nom + " " + self.prenom + " J'ai " + self.age + " ans")
        

eleve1=Eleve("Calbuth" , "Jean" , "17")
eleve2=Eleve("Lagaffe" , "Paul" , "16")
eleve1.presentation()
print("ou alors :")
Eleve.presentation(eleve2)
print(eleve1)
print(eleve2)
print(Eleve)
print(id(Eleve))
print(hex(id(Eleve)))
print(eleve1.nom)
print(isinstance(eleve1,Eleve))
```

@lesdecorateurs <br> Le principe de la décoration en Python est d’appliquer un décorateur à une fonction, afin de retourner un nouvel objet (généralement une fonction). On peut donc voir le décorateur comme une fonction prenant une fonction en paramètre, et retournant une nouvelle fonction

```python
def plusUn(nombre):
    return int(nombre) + 1

def plusDeux(nombre):
    return int(nombre) + 2

ajouterUn = plusUn

ajouterUn(10)

```

```python
def mon_decorateur(fonction):
    def nouvelle_fonction(*args, **kwargs):
        print("Avant l’appel")
        resultat = fonction(*args, **kwargs)
        print("Après l’appel")
        return resultat
    return nouvelle_fonction

@mon_decorateur
def dire_bonjour(nom):
    print(f"Bonjour {nom} !")

dire_bonjour("Alice")
```

En POO les classes filles héritent des propriétés et méthodes des classes mères. Une classe fille peut hériter de plusieurs classes mères. Hériter veut dire avoir accès aux méthodes et aus attributs des classes mères.

```python
class Mere1:
    pass


class Mere2:
    pass


class Fille1 (Mere1):
    # heritage
    pass


class Fille2 (Mere1, Mere2):
    # heritage multiple
    pass
 
    
    
print (issubclass(Fille1, Mere2))
print (issubclass(Fille1, Mere1))
print (issubclass(Fille2, Mere1))
```

```python
class Voiture:
    # constructeur
    def __init__(self,nom = "clio",marque= "renault"):
        # attributs d'instance
        self.nom = nom
        self.marque = marque
        
    # méthode
    def identite(self):
        print(self.nom , self.marque)
        
        
class Voiture_de_sport(Voiture):
    # constructeur
    def __init__(self,nom = "clio",marque= "renault", puissance = 200):
        # héritage de la classe Voiture
        Voiture.__init__(self, nom, marque)
        # attribut d'instance
        self.puissance = puissance
        
    # méthode
    def performance(self):
        print(self.nom , self.marque)
        print(self.puissance)
        

porsche = Voiture_de_sport("porsche", "boxter", 240)

porsche.performance()

titine = Voiture()
titine.identite()
        
        
    
```

```python
class Option :
    def __init__(self,element):
        self.element=element
        
    def __add__(self,un_objet):
        return Option(str(self.element) + " et "+ un_objet.element)
    
    def affichage(self):
        print(self.element)
        
option1=Option("Vitre electrique")
option2=Option("direction asistée")
option3=Option("Sieges en cuir")

option2.affichage()

toutes_options = option1 + option2 + option3
toutes_options.affichage()
```

<b>Le polymorphisme</b> <br> L’héritage et le polymorphisme forment avec le principe d’encapsulation les trois piliers de la programmation orientée objet. “Polymorphisme” signifie littéralement “plusieurs formes”. Dans le contexte de la programmation orientée objet, le polymorphisme est un concept qui fait référence à la capacité d’une variable, d’une fonction ou d’un objet à prendre plusieurs formes, c’est-à-dire à sa capacité de posséder plusieurs définitions différentes.

```python
class Animaux:
    def __init__(self, nom):
        self.animal_nom = nom
        
    def seNourrir(self):
        print("Je mange des légumes")
        
    def seDeplacer(self):
        print("Je marche avec mes pieds")
        

class Chien(Animaux):
    def seNourrir(self):
        print("Je mange des croquettes")
        
    def seDeplacer(self):
        print("je cours avec mes pattes")
        

class Oiseau(Animaux):
    def seNourrir(self):
        print("Je mange du planqueton")
        
    def seDeplacer(self):
        print("je nage")
        
        

animal1 = Chien("Labrador")
animal1.seDeplacer()
animal1.seNourrir()
print(animal1.animal_nom)
        
```
