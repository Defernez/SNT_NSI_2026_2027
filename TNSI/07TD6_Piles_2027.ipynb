<p dir="auto" data-sourcepos="9:1-9:192">La <strong>pile</strong> est une structure de données qui fonctionne sur le principe du <strong>dernier arrivé premier sorti</strong>. Pour une pile, on parle de mode <strong>LIFO</strong> pour <strong>L</strong>ast <strong>I</strong>n, <strong>F</strong>irst <strong>O</strong>ut.</p>

<strong>Nous allons utiliser la structure chaînée</strong> pour implémenter notre Pile. Ce n'est pas la seule façon de faire. On pourrait, par exemple, utiliser la structure  des tableaux de Python. Mais la structure chaînée est bien plus adaptée à l'ajout et au retrait d'éléments, particulièrement en tête,  que celle en tableau indexé comme nous l'avons vu en étudiant les listes chaînées.

```python
class Maillon:
    """
    Représente un maillon (ou un nœud) d'une liste chaînée.

    Attributs:
        valeur: La valeur stockée dans le maillon.
        suivant: Le maillon suivant dans la liste chaînée (par défaut, None si aucun suivant).

    Exemples:
        >>> m1 = Maillon(5)
        >>> m1.valeur
        5
        >>> m1.suivant is None
        True

        >>> m2 = Maillon(10)
        >>> m1.suivant = m2
        >>> m1.suivant.valeur
        10
    """

    def __init__(self, valeur, suivant=None):
        """
        Initialise un maillon de la liste chaînée.

        Arguments:
            valeur: La valeur à stocker dans ce maillon.
            suivant: Le maillon suivant dans la chaîne, par défaut None (pas de suivant).

        Exemples:
            >>> m = Maillon(3)
            >>> m.valeur
            3
            >>> m.suivant is None
            True

            >>> suivant_maillon = Maillon(4)
            >>> m = Maillon(3, suivant_maillon)
            >>> m.suivant.valeur
            4
        """
        self.valeur = valeur  # La valeur stockée dans le maillon.
        self.suivant = suivant  # Référence vers le maillon suivant, None par défaut.
```
