En Python, les dictionnaires sont des structures de données très puissantes basées sur une association clé-valeur. Cela signifie que chaque élément du dictionnaire est composé d'une clé unique et d'une valeur associée.

```python
mon_dictionnaire = {
    "nom": "Alice",
    "age": 30,
    "ville": "Paris"
}
```

nom, age, ville sont les clés.

"Alice", 30, "Paris" sont les valeurs associées.

```python
# Accéder à une valeur
print(mon_dictionnaire["nom"])  # Affiche "Alice"
```

```python
# Ajouter ou modifier une paire clé-valeur
mon_dictionnaire["profession"] = "Ingénieur"
mon_dictionnaire["age"] = 31  # Modification
print(mon_dictionnaire) 
```

```python
# Supprimer une clé
del mon_dictionnaire["ville"]
print(mon_dictionnaire) 
```

```python
# Parcourir un dictionnaire
# Clés
for cle in mon_dictionnaire:
    print(cle)
```

```python
# Parcourir un dictionnaire
# Valeurs
for valeur in mon_dictionnaire.values():
    print(valeur)
```

```python
# Parcourir un dictionnaire
# Clés et valeurs
for cle, valeur in mon_dictionnaire.items():
    print(cle, valeur)
```

Méthodes utiles

```python
mon_dictionnaire.keys()     # Retourne les clés
```

```python

mon_dictionnaire.values()   # Retourne les valeurs
```

```python
mon_dictionnaire.items()    # Retourne les paires clé-valeur
```

```python
mon_dictionnaire.get("nom") # Retourne la valeur ou None si la clé n'existe pas
```

```python
mon_dictionnaire.clear()    # Vide le dictionnaire
```

```python
print(mon_dictionnaire)
```

🔐 Remarques

    Les clés doivent être immutables (ex : chaînes, nombres, tuples).

    Les dictionnaires ne conservent pas d’ordre avant Python 3.7. Depuis, ils conservent l’ordre d’insertion.

```python
contacts = {
    "Alice": "alice@gmail.com",
    "Bob": "bob@yahoo.com",
    "Claire": "claire@hotmail.com"
}

# Afficher l'email de Bob
print("Email de Bob :", contacts["Bob"])

# Ajouter un nouveau contact
contacts["David"] = "david@protonmail.com"

# Modifier l'email de Claire
contacts["Claire"] = "claire@outlook.com"

# Supprimer Alice
del contacts["Alice"]

# Afficher tous les contacts
for nom, email in contacts.items():
    print(nom, ":", email)
```

###  Conclusion 
Les dictionnaires sont très utiles quand on a besoin d'ajouter ou de supprimer des éléments dans n'importe quel ordre. Ils remplacent avantageusement les listes lorsqu’il s’agit de traiter des ensembles de données numérotées, dont les numéros ne se suivent pas
