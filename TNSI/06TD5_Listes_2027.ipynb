<p dir="auto" data-sourcepos="123:1-123:135">Lorsqu'un informaticien parle de "liste", il ne parle en réalité pas des tableaux comme vu ci-dessus, mais des <strong>listes chaînées</strong>.</p>
<h3 dir="auto" data-sourcepos="125:1-125:15">
<a aria-hidden="true" href="#1-principe-1" class="anchor" id="user-content-1-principe-1"></a>1. Principe</h3>
<p dir="auto" data-sourcepos="127:1-127:190">La structure chaînée d'une liste n'utilise pas un tableau mais une série de <strong>maillons</strong> chacun composé d'une valeur et d'un lien vers le maillon suivant, le dernier maillon étant vide.</p>

![imageListeChain](https://upload.wikimedia.org/wikipedia/commons/thumb/6/6d/Singly-linked-list.svg/1920px-Singly-linked-list.svg.png)

<h3 dir="auto" data-sourcepos="143:1-143:34">

<p dir="auto" data-sourcepos="145:1-145:153">Cette structure est particulièrement performante pour insérer une cellule n'importe où (*<em>et particulièrement en tête</em>) car il suffit pour cela de :</p>
<ul dir="auto" data-sourcepos="147:1-160:0">
<li data-sourcepos="147:1-154:0">
<p data-sourcepos="147:3-147:29">créer un nouveau maillon :</p>

![img](https://upload.wikimedia.org/wikipedia/commons/0/01/Liste_ajout1.png)

 . Modifier les liens entre les maillons :

![img](https://upload.wikimedia.org/wikipedia/commons/a/a1/Liste_ajout3.png)

Il n'y a donc pas besoin de décaler tous les autres maillons, comme c'était le cas pour un tableau : <strong>le coût est faible et constant, quelque soit la taille de la liste chaînée</strong>.</p>
<p dir="auto" data-sourcepos="163:1-163:295">En revanche, pour <em>atteindre</em> un maillon particulier de la liste, il faudra parcourir tous les maillons en partant du premier et <em>remonter</em> la liste jusqu'au maillon désiré. L'accès à un maillon particulier via son indice dans la liste est donc beaucoup plus coûteux que pour les tableaux !</p>

<h2 dir="auto" data-sourcepos="225:1-225:44">
<a aria-hidden="true" href="#iv-impl%C3%A9mentation-dune-liste-cha%C3%AEn%C3%A9e" class="anchor" id="user-content-iv-implémentation-dune-liste-chaînée"></a>IV. Implémentation d'une liste chaînée</h2>
<p dir="auto" data-sourcepos="227:1-227:230">Pour l'instant, nous avons travaillé uniquement avec l'<em>interface</em> des listes chaînées, c'est-à-dire à la main, avec des schémas et avec l'ensemble des opérations que l'on peut réaliser (insertion, accès, suppression...).</p>
<p dir="auto" data-sourcepos="229:1-229:70">Nous allons maintenant <em>implémenter</em> nos listes chaînées en Python.</p>

```python
class Cellule:
    """une cellule d'une liste chaînée
        elle comprend 2 attributs 
        v : valeur 
        s : lien vers la cellule suivante
    """
    def __init__(self, v, s):
        self.valeur = v
        self.suivant = s

class ListeOrdo:
    def __init__(self):
        self.cellule =None 
    
    def est_vide(self):
        return self.cellule is None 
    
    def ajouter(self,v):
        c=self.cellule
        if v <= c.valeur:
            c = Cellule(v,c)
            self.cellule =c
        while v > c.valeur :
            if c.suivant is not None and c.suivant.valeur <v :
                c = c.suivant
            else :
                c.suivant = Cellule(v,c.suivant)
                c= c.suivant
    
                
                
    def compter(self,v):
        compt=0 #initilaisation du compteur
        c=self.cellule
        if self.est_vide() == False :
            while v >= c.valeur :
                if c.valeur == v:
                    compt +=1
                c=c.suivant
        return compt
    
    
     
    def __str__ (self):
        if self.est_vide(): 
            return "nil"
        else : 
            c =self.cellule
            res =str(c.valeur)
            while c.suivant is not None :
                c=c.suivant
                res += "-> " + str(c.valeur)
            return res
                
        
        
```
